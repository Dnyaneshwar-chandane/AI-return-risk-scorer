```python
import pandas as pd
import numpy as np
data=pd.read_csv('../data/raw/orders.csv',parse_dates=['order_date'])
```

```python
data['order_dayofweek']=data['order_date'].dt.dayofweek
```

```python
data['is_weekend']=(data['order_dayofweek'] >= 5).astype(int)
```

```python
data['is_weekend'].value_counts()
```

```python
data['order_month']=data['order_date'].dt.month
```

```python
data = data.sort_values(
    ["customer_id", "order_date"]
)
```

```python
data["time_since_last_order"]=data.groupby('customer_id')['order_date'].diff().dt.days
```

```python
data["time_since_last_order"].isnull().mean()*100
```

```python
data.info()
```

```python
data["discount_amount"] = (
    data["order_value"] * data["discount_pct"] / 100
)
```

```python
data["net_order_value"] = (
    data["order_value"] * (1 - data["discount_pct"] / 100)
)
```

```python
data["delivery_delay_ratio"] = (
    data["delivery_days"] /
    data["expected_delivery_days"]
)
```
